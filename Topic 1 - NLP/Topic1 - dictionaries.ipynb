```python
f = open("animals.txt", "r")
data = f.read()
type(data)
data
f.close()
animals = data.split(",")
counter = {}
for animal in animals:
    if animal in counter:
        counter[animal] +=1
    else:
        counter[animal] = 1
counter
```
