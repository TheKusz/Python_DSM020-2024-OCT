```python
from urllib.request import urlopen
from bs4 import BeautifulSoup
html = urlopen('https://www.doc.gold.ac.uk/~smcgr004/')
# Let's scrape my website, because it's the only place I can guarantee won't change!
bs = BeautifulSoup(html, "html.parser")
```

```python
# Here we want to look for <div> level elements that conform to a particular class. These are often defined in CSS
# as something.SomethingElse https://www.w3schools.com/cssref/sel_class.asp
container = bs.findAll('div', {'class': 'container-content'})
for name in container:
    print(name.get_text())
```

```python
# There is some guff in the previous example that we might want to clean up. You can either do it using the NLP
# techniques or throw some regex at it https://www.w3schools.com/python/python_regex.asp


# Let's grab the 2nd largest headings <h2> elements
headings = bs.find_all(['h2'])
print([heading for heading in headings])

# for heading in headings:
#    print(heading.text)
```

```python
# We might want to grab all images on the page. One approach is this way.
images = bs.findAll('img')
images
```

```python
# The missing alt text is really bad practice. This is what people using a screen reader would see, so I really should
# have made an effort to include an accurate description of the images.

# Let's iterate and get just the names of the files then.
for image in images:
    print(image['src'])
```

```python
# Let's do something a bit more complicated. Let's create a mini web spider to grab all of the links.
from urllib.request import urlopen
from bs4 import BeautifulSoup 

html = urlopen('https://www.doc.gold.ac.uk/~smcgr004/')
bs = BeautifulSoup(html, 'html.parser')

cleaned_list = list()
for link in bs.find_all('a'):
    if 'href' in link.attrs:
        print(link.attrs['href'])
        if "mailto" not in link.attrs['href']:
            cleaned_list.append(link.attrs['href'])
# Notice something? We get *all* anchor elements including those used for navigating within the page.
# We've also got some mailto links in here, probably not useful. Can you process these to remove unnecessary output?

cleaned_list
```
