### Do cats reduce stress?
Let's search for the answer in the following article!

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("https://psychcentral.com/health/your-cat-depression-anxiety")
bs = BeautifulSoup(html, "html.parser")
```

```python
## Let's see the structure

for head1 in bs.findAll('h1'):
    print(head1.get_text())
    for head2 in head1.find_all_next(['h2']):
        print("\t" + head2.get_text())
```

```python
## Check the images
for i in bs.findAll('img'):
    print(i['src'])
```

### Webspider
Script that browses automatically the web by using the different links from webpages.

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("https://psychcentral.com/health/your-cat-depression-anxiety")
bs = BeautifulSoup(html, 'html.parser')

for link in bs.find_all('a'):
    if 'href' in link.attrs:
        # None
        print(link.attrs['href'])
```

### Let's find information in the article

```python
import re
# Search for the world help, and find the context around it

for i in bs.find_all('p', string=re.compile('help')):
    print(i.get_text())
    print("-----")
```

```python
# Let's see what are the different classes in the html
# There were lots of css, so I removed them

html_classes = set()
for tag in bs.find_all(True):
    if tag.has_attr('class'):
        html_classes.update(c for c in tag['class'] if 'css' not in c)

html_classes
```

```python
## Nothing interesting, so I will try to search for research to learn more
for i in bs.find_all('p', string=re.compile('research')):
    print(i.get_text())
    print("-----")
```
