### Clean the data

```python
import pandas as pd
df = pd.read_csv('data/JIRA_data_shortened.csv')
```

```python
## comment fields
df.drop(columns=df.filter(like='Comment', axis = 1).columns, inplace=True)
for i in df.filter(like='Comment', axis = 1).columns:
    print(i)
```

```python
single_value_columns = df.columns[df.nunique(dropna=True) <= 1]
del_custom = [col for col in single_value_columns if 'Custom field' in col]
df.drop(columns=del_custom, inplace=True)
del_test = [col for col in df.columns if 'Test' in col]
df.drop(columns=del_test, inplace=True)
```

```python
df.to_csv('data/JIRA_data_shortened.csv')
```
