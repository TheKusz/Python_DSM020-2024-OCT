```python
# First, import packages and load some data
import pandas      as pd
import numpy       as np
import scipy       as sp
import scipy.stats as stats

datafile = './pima-indians-diabetes.csv'

df = pd.read_csv(datafile, header=0, index_col=None)
df.describe()
```

```python
# Check if there is any NaN
df.isna().sum()
```

```python
df_nan = df.copy()

# Convert to float to avoid type mismatch with NaN
df_nan.iloc[:, 1:8] = df_nan.iloc[:, 1:8].astype(float)

# Replace 0 to NAN
df_nan.iloc[:, 1:8] = df_nan.iloc[:, 1:8].replace({0: np.nan})
```

```python
df_nan
```

```python
# Check counf of NaN values / columns
df_nan.isna().sum()
```

```python
# Check count of 0 values / columns
(df_nan == 0).sum()
```

```python
def print_stats(df):
    for col in df.columns:
        col_min  = np.min(df[col])
        col_max = np.max(df[col])
        col_mean = np.nanmean(df[col])
        col_median = np.nanmedian(df[col])
        freq = stats.mode(df[col])
        print(f"{col}:\nmin:{col_min}, max:{col_max}, mean:{col_mean}, median:{col_median}, mode:{freq.mode} count:{freq.count}\n")
```

```python
print_stats(df_nan)
```

```python
# Hint: centre([0,1,2,3,5,10]) => [-1.059, -0.757, -0.454, -0.151, 0.454, 1.967]
def centre_population(v):
    v = (v - np.nanmean(v)) / np.nanstd(v)
    return v

# Let's see if there is any difference if we calculate the std with the sample's formula (N-1)
def centre_sample(v):
    v = (v - np.nanmean(v)) / np.nanstd(v, ddof=1)
    return v

# test
centre_population([0,1,2,3,5,10])
```

```python
# Let's make a copy of df, to hold the centred values:
cdf_pop = df_nan.copy()
cdf_sample = df_nan.copy()
```

```python
# Now fill in cdf by applying your centre() function to each column in df
cdf_pop = cdf_pop.apply(centre_population, axis=0)
cdf_sample = cdf_sample.apply(centre_sample, axis=0)
```

```python
# Finally! Let's look for outliers by getting the min and max in the new centred df
# Let's see if there is any difference in the result using population / sample std
print_stats(cdf_pop)
print("----------------")
print_stats(cdf_sample)
```
